## Step 4 — Generate the Data Profiling Report

Now that the dataset has been successfully loaded and the required attributes have been verified, generate an automated profiling report.

The objective is to collect evidence about the current state of the data.

The profiling report does **not** decide which values are correct, which records should be cleaned, or which validation rules should be implemented.

### 4.1 Import the Profiling Tool

```python
from data_profiling import ProfileReport
```

FG Data Profiling generates an extended analysis of a pandas DataFrame, including descriptive statistics, missing values, distributions, cardinality, correlations, and profiling warnings.

### 4.2 Create the Profiling Report

```python
profile = ProfileReport(
    df,
    title="Candidate Applications — Data Profiling Report",
)
```

Generating the report may take some time depending on the computer and dataset size.

Wait until the profiling process is complete before continuing.

### 4.3 Display the Report

```python
profile.to_notebook_iframe()
```

Inspect the report carefully.

Pay particular attention to:

- missing values;
- variable types;
- minimum and maximum values;
- numeric distributions;
- categorical frequencies;
- unique values and cardinality;
- duplicate-related information;
- unusual or suspicious patterns;
- warnings reported by the profiling tool.

Do not make cleaning or validation decisions yet.

### 4.4 Export the Profiling Report

Save the report as an HTML artifact in the project's `reports/` folder.

```python
REPORT_PATH = (
    PROJECT_ROOT
    / "reports"
    / "candidates_profiling.html"
)

REPORT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)

profile.to_file(
    REPORT_PATH
)

print("Profiling report saved to:")
print(REPORT_PATH)
```

The generated report should now be available at:

`reports/candidates_profiling.html`

This file is a profiling artifact that can be reviewed without regenerating the report.
